```python
### simple a simple txt file

### created a directory inside data folder
import os
os.makedirs("../data/text_files",exist_ok=True)
```

```python
sample_texts = {
    "../data/text_files/python_intro.txt": """python programming introduction 
    
key features:
easy to read,
interpreted language,
multi-paradigm,
cross-platform,
rich-ecosystem"""
}

for file_path,content in sample_texts.items():
    os.makedirs(os.path.dirname(file_path), exist_ok=True)
    with open(file_path, "w", encoding="utf-8") as file:
        file.write(content)

print("sample text files created")
```

```python
### text loader

from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/python_intro.txt", encoding="utf-8")
loader  
document=loader.load()
```

```python
## now want to print the data in the document
print(document)
```

```python
### Directory loader , like reading files , all files from directory

from langchain_community.document_loaders import DirectoryLoader

### load all text files from the directory
dir_loader = DirectoryLoader(
    "../data/text_files", 
glob="**/*.txt", ## which kind of files you want to load from the directory
loader_cls=TextLoader, ## which file you're planning to load like text, pdf, word
loader_kwargs={"encoding": "utf-8"},
show_progress=False
)

documents=dir_loader.load()
documents
```

```python
### for pdf document loader

from langchain_community.document_loaders import DirectoryLoader,PyPDFLoader,PyMuPDFLoader

### Load all PDF files from the directory

dir_loader = DirectoryLoader(
    "../data/resume",
    glob="**/*.pdf",
    loader_cls=PyMuPDFLoader,
    show_progress=False
)

pdf_documents = dir_loader.load()
pdf_documents
```
